```python
#Scalar multiplication with function
y = np.sin(x)
y1 = 2 * np.sin(x)
y2 = 0.5 * np.sin(x)

plt.plot(x,y, label = " y = sin(x)", color = "blue")
plt.plot(x,y1, label = " y = 2 * sin(x)", color = "red")
plt.plot(x,y2, label = " y = 0.5 * sin(x)", color = "green")

plt.axvline(0, color = "black", linewidth = 1.5, linestyle = "--")
plt.axhline(0, color = "black", linewidth = 1.5, linestyle = "--")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of sin(x),k 2*sin(x),0.5*sin(x)")
plt.grid()
plt.legend()

plt.show()
```

```python
#Scalar multiplication with function
y3 = -2 * np.sin(x)

plt.plot(x,y1, label = " y = 2 * sin(x)", color = "blue")
plt.plot(x,y3, label = " y = -2 * sin(x)", color = "red")

plt.axvline(0, color = "black", linewidth = 1.5, linestyle = "--")
plt.axhline(0, color = "black", linewidth = 1.5, linestyle = "--")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of (y = 2* sin(x) vs y = -2* sin(x))")
plt.grid()
plt.legend()
plt.show()
```

```python
#Scalar addition with function
# range will change from [-1, +1] to [-1+2, +1+2]
y = np.sin(x)
y4 = 2 + np.sin(x)

plt.plot(x,y, label = " y = sin(x)", color = "blue")
plt.plot(x,y4, label = " y = 2 + sin(x)", color = "red")

plt.axvline(0, color = "black", linewidth = 1.5, linestyle = "--")
plt.axhline(0, color = "black", linewidth = 1.5, linestyle = "--")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of sin(x) and 2+ sin(x)")
plt.grid()
plt.legend()

plt.show()
```

```python
#Scalar multiplication with x
# range will not change

y = np.sin(x)
y5 = np.sin(2*x)

plt.plot(x,y, label = " y = sin(x)", color = "blue")
plt.plot(x,y5, label = " y = sin(2*x)", color = "red")

plt.axvline(0, color = "black", linewidth = 1.5, linestyle = "--")
plt.axhline(0, color = "black", linewidth = 1.5, linestyle = "--")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of sin(x) and sin(2*x)")
plt.grid()
plt.legend()

plt.show()
```

```python
#Scalar addition with x
y = np.sin(x)
y6 = np.sin(2+x)

plt.plot(x,y, label = " y = sin(x)", color = "blue")
plt.plot(x,y6, label = " y = sin(2+x)", color = "red")

plt.axvline(0, color = "black", linewidth = 1.5, linestyle = "--")
plt.axhline(0, color = "black", linewidth = 1.5, linestyle = "--")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of sin(x) and sin(2+x)")
plt.grid()
plt.legend()

plt.show()
```
