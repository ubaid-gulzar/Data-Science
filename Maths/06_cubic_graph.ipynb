```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,15,1000)
y = np.power(x,3) - 6 * np.square(x) + 9 * x + 1
```

```python
#graph of y = x^3 - 6x^2 + 9x +1
plt.plot(x,y,color = "blue")

plt.axvline(0, color = "black", linewidth = 1.5, linestyle = "--")
plt.axhline(0, color = "black", linewidth = 1.5, linestyle = "--")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of y = x^3 - 6x^2 + 9x +1")
plt.grid()
plt.legend()
plt.ylim(-10,10)
plt.xlim(-5,5)
plt.show()
```
