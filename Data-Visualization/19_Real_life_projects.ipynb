```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
fig,ax = plt.subplots(2)
for i in range(2):
    sns.lineplot(
        data = tips,
        x = "day",
        y = "total_bill",
        hue = "sex",
        ax = ax[i]
    )
    ax[i].set_title("Days vs Total_Bill")
fig.tight_layout()
```
