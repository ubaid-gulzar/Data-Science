```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sns.set_theme()
```

```python
flights = sns.load_dataset("flights")
print(flights)
```

```python
flights_pivot = flights.pivot(index = "month",columns = "year",values= "passengers")
print(flights_pivot)
```

```python
sns.heatmap(
    flights_pivot,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
)
plt.title("Number of passengers in flights in month")
plt.tight_layout()
```
